Lab 4: Student Performance Classifier

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

np.random.seed(42)
n_samples = 200

fail_students = np.random.normal(loc=[50, 45, 40, 3], scale=[8, 8, 8, 1.5], size=(n_samples // 3, 4))
average_students = np.random.normal(loc=[70, 65, 65, 6], scale=[8, 8, 8, 1.5], size=(n_samples // 3, 4))
excellent_students = np.random.normal(loc=[90, 88, 85, 10], scale=[6, 6, 6, 1.5], size=(n_samples // 3, 4))

X = np.vstack([fail_students, average_students, excellent_students])
y = np.array([0]*len(fail_students) + [1]*len(average_students) + [2]*len(excellent_students))

X[:, :3] = np.clip(X[:, :3], 0, 100)
X[:, 3] = np.clip(X[:, 3], 0, 20)

print("Feature shape:", X.shape)
print("Classes:", np.unique(y))

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)
y_categorical = to_categorical(y_encoded)

print(y_categorical[:5])

X_train, X_test, y_train, y_test = train_test_split(
    X, y_categorical, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = Sequential([
    Dense(16, input_dim=4, activation='relu'),
    Dense(8, activation='relu'),
    Dense(3, activation='softmax')   # 3 output neurons, one per class
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()

history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=8,
    validation_split=0.1,
    verbose=1
)

loss, accuracy = model.evaluate(X_test, y_test)
print(f"\nTest Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")

class_names = ["Fail", "Average", "Excellent"]
predictions = model.predict(X_test)
predicted_classes = np.argmax(predictions, axis=1)
actual_classes = np.argmax(y_test, axis=1)

print("\nPredicted classes:", [class_names[i] for i in predicted_classes[:10]])
print("Actual classes:   ", [class_names[i] for i in actual_classes[:10]])
```
