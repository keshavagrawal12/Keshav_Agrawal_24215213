Lab 2: Spam Email Detector

```python
import tensorflow as tf
import numpy as np

# Features: [suspicious_link (1=yes), unknown_sender (1=yes)]
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
# Email is SPAM only if BOTH suspicious link AND unknown sender are present
y = np.array([[0], [0], [0], [1]], dtype=np.float32)

model = tf.keras.Sequential([
    tf.keras.layers.Dense(4, input_dim=2, activation='relu'),   
    tf.keras.layers.Dense(1, activation='sigmoid')               
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

labels = ["Safe link, Known sender", "Safe link, Unknown sender",
          "Suspicious link, Known sender", "Suspicious link, Unknown sender"]

print("Predictions BEFORE training:")
before = model.predict(X, verbose=0)
for i, label in enumerate(labels):
    print(f"{label:32s} -> confidence: {float(before[i][0]):.4f} (random, untrained)")

model.fit(X, y, epochs=10, verbose=0)

print("\nPredictions AFTER 10 epochs:")
after = model.predict(X, verbose=0)
for i, label in enumerate(labels):
    prob = float(after[i][0])
    verdict = "SPAM" if prob >= 0.5 else "NOT SPAM"
    print(f"{label:32s} -> {verdict} (confidence: {prob:.4f})")
```
