Lab 3: Smart irrigation system

```python
import torch
import torch.nn as nn
import torch.optim as optim
torch.manual_seed(0) 

# Features: [soil_is_dry (1=yes), no_rain_forecast (1=yes)]
X = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [0], [0], [1]], dtype=torch.float32)

class SprinklerNN(nn.Module):
    def __init__(self):
        super(SprinklerNN, self).__init__()
        self.fc1 = nn.Linear(2, 4)  
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(4, 1)   
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.relu(self.fc1(x))
        out = self.sigmoid(self.fc2(out))
        return out

model = SprinklerNN()
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(100):
    outputs = model(X)              # Step 1: make a guess
    loss = criterion(outputs, y)    # Step 2: check how wrong the guess was

    optimizer.zero_grad()           # Step 3: clear old corrections
    loss.backward()                 # Step 4: backpropagation - figure out what to fix
    optimizer.step()                # Step 5: apply the fix (update weights)

labels = ["Wet soil, Rain expected", "Wet soil, No rain expected",
          "Dry soil, Rain expected", "Dry soil, No rain expected"]

print("Smart Irrigation Predictions (after 100 epochs):")
with torch.no_grad():
    predictions = model(X)
    for i, label in enumerate(labels):
        prob = float(predictions[i])
        verdict = "SPRINKLER ON" if prob >= 0.5 else "SPRINKLER OFF"
        print(f"{label:32s} -> {verdict} (confidence: {prob:.4f})")
```
