Lab 3: Spam Detector

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Features: [suspicious_link (1=yes), unknown_sender (1=yes)]
X = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [0], [0], [1]], dtype=torch.float32)

class SpamDetectorNN(nn.Module):
    def __init__(self):
        super(SpamDetectorNN, self).__init__()
        self.fc1 = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(4, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.relu(self.fc1(x))
        out = self.sigmoid(self.fc2(out))
        return out

model = SpamDetectorNN()
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Train for 100 epochs this time -- long enough to actually converge
for epoch in range(100):
    outputs = model(X)
    loss = criterion(outputs, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

labels = ["Safe link, Known sender", "Safe link, Unknown sender",
          "Suspicious link, Known sender", "Suspicious link, Unknown sender"]

print("PyTorch Spam Detection (after 100 epochs):")
with torch.no_grad():
    predictions = model(X)
    for i, label in enumerate(labels):
        prob = float(predictions[i])
        verdict = "SPAM" if prob >= 0.5 else "NOT SPAM"
        print(f"{label:32s} -> {verdict} (confidence: {prob:.4f})")
```
